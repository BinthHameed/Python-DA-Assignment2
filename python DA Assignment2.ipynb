```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


pd.set_option('display.max_columns', None)
```

```python
# Load the taxis dataset
df = sns.load_dataset("taxis")

# Display the first five rows
df.head()
```

```python
# Display the first five records
display(df.head())
```

```python
# Display the last five records
display(df.tail())
```

```python
# Check the number of rows and columns
print("Dataset Shape:", df.shape)
```

```python
# Check column names and data types
df.info()
```

```python
# Generate statistical summary
display(df.describe())
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Displaying only the columns with missing values
missing_values = df.isnull().sum()

missing_columns = missing_values[missing_values > 0]

print("Columns with missing values:")
print(missing_columns)
```

```python
# Convert timestamp columns into datetime format

df['pickup'] = pd.to_datetime(df['pickup'], errors='coerce')
df['dropoff'] = pd.to_datetime(df['dropoff'], errors='coerce')
```

```python
# Remove rows with missing pickup timestamp or fare

df.dropna(subset=['pickup', 'fare'], inplace=True)
```

```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

# Fill missing numerical values with median
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Identify categorical columns
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

# Fill missing categorical values with mode
for col in categorical_cols:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("Missing values after treatment:")
print(df.isnull().sum())
```

```python
# Visualization
line_data = df.sort_values('pickup')

# Create line chart
plt.figure(figsize=(12, 5))

plt.plot(line_data['pickup'], line_data['fare'])

plt.title('Taxi Fare Over Time')
plt.xlabel('Pickup Timestamp')
plt.ylabel('Fare Amount')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Group data by pickup borough and calculate total fare
borough_fare = df.groupby('pickup_borough')['fare'].sum()

# Create bar chart
plt.figure(figsize=(9, 5))

borough_fare.plot(kind='bar')

plt.title('Total Fare by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Pie chart
# Count trips by payment method
payment_counts = df['payment'].value_counts()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Trips by Payment Method')
plt.show()
```

```python
#Histogram
# Create histogram
plt.figure(figsize=(9, 5))

plt.hist(df['distance'], bins=20, edgecolor='black')

plt.title('Distribution of Taxi Trip Distances')
plt.xlabel('Distance')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Box plot
# Create box plot
plt.figure(figsize=(10, 6))

df.boxplot(
    column='tip',
    by='pickup_borough',
    grid=False
)

plt.title('Distribution of Tips by Pickup Borough')
plt.suptitle('')
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Count plot- no. of trips by pickup borough
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='pickup_borough'
)

plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter plot- relation bw distance and fare
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='distance',
    y='fare',
    hue='pickup_borough'
)

plt.title('Relationship Between Distance and Fare')
plt.xlabel('Trip Distance')
plt.ylabel('Fare Amount')

plt.tight_layout()
plt.show()
```

```python
# Heat map- correlation between  numerical variable
# Select numerical columns
numeric_data = df[['distance', 'fare', 'tip', 'tolls', 'total']]

# Calculate correlation matrix
correlation_matrix = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Taxi Dataset')
plt.tight_layout()
plt.show()
```

```python
# Pair plot
# Identify the five most common pickup zones
top_zones = df['pickup_zone'].value_counts().head(5).index

# Filter dataset for these zones
pair_data = df[df['pickup_zone'].isin(top_zones)]

# Create pair plot
sns.pairplot(
    pair_data.sample(min(1000, len(pair_data)), random_state=42),
    vars=['distance', 'fare', 'tip', 'total'],
    hue='pickup_zone',
    diag_kind='hist'
)

plt.show()
```

```python
# Violin plot -fare distribution by payment type
plt.figure(figsize=(9, 6))

sns.violinplot(
    data=df,
    x='payment',
    y='fare'
)

plt.title('Fare Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare Amount')

plt.tight_layout()
plt.show()
```
